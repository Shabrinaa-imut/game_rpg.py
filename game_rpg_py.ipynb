```python
class Hero:

    def __init__(self, nama, hp_awal, attack_power):
        self.nama = nama
        self.attack_power = attack_power
        # ENKAPSULASI: HP dibuat Private agar aman dari peretasan/akses langsung
        self.__hp = hp_awal

    # GETTER: Membaca nilai HP secara aman
    def get_hp(self):
        return self.__hp

    # SETTER: Mengubah nilai HP dengan logika validasi
    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def info(self):
        print(
            f"Hero: {self.nama} | HP: {self.get_hp()} | Power:"
            f" {self.attack_power}"
        )

    # INTERAKSI ANTA ROBJEK: Parameter 'lawan' menerima objek utuh
    def serang(self, lawan):
        print(f"{self.nama} menyerang {lawan.nama}!")
        lawan.diserang(self.attack_power)

    def diserang(self, damage):
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(
            f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}"
        )


class Mage(Hero):

    def __init__(self, nama, hp_awal, attack_power, mana):
        # Memanggil constructor milik Parent (Hero)
        super().__init__(nama, hp_awal, attack_power)
        self.mana = mana

    def info(self):
        print(
            f"{self.nama} (Mage) | HP: {self.get_hp()} | Mana: {self.mana}"
        )

    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"{self.nama} menggunakan Fireball ke {lawan.nama}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2)  # Damage 2x lipat
        else:
            print(f"{self.nama} gagal skill! Mana tidak cukup.")


if __name__ == "__main__":
    print("--- Latihan 1 & 2: Instansiasi & Serangan ---")
    hero1 = Hero("Layla", 100, 15)
    hero2 = Hero("Zilong", 120, 20)

    hero1.info()
    hero2.info()

    print("\n--- Pertarungan Dimulai ---")
    hero1.serang(hero2)
    hero2.serang(hero1)

    print("\n--- Latihan 3: Pewarisan (Mage) ---")
    balmond = Hero("Balmond", 200, 10)
    eudora = Mage("Eudora", 80, 30, 100)

    eudora.info()
    eudora.serang(balmond)
    eudora.skill_fireball(balmond)

    print("\n--- Latihan 4: Pengujian Enkapsulasi & Validasi ---")
    hero1.set_hp(-50)  # Dicoba set negatif, akan ditahan di angka 0
    print(f"HP Layla setelah diset negatif: {hero1.get_hp()}")
```

Analisis 1
- Akses Publik: Mengubah hero1.hp = 500 secara langsung mengubah nilai atributnya. Hal ini terjadi karena pada awal latihan, atribut hp masih bersifat publik (public attribute) sehingga bebas dimodifikasi dari luar class.  

Analisis 2
- Passing Object: Parameter lawan perlu menerima objek utuh (bukan sekadar string nama) agar class Hero dapat secara langsung membaca variabel milik lawan (lawan.nama) sekaligus memicu method milik lawan tersebut (lawan.diserang(...)).  

Analisis 3
- Penyebab Error tanpa super(): Jika super().__init__() dihapus/dijadikan komentar, muncul AttributeError saat memanggil eudora.info().  Penjelasan: Tanpa super(), proses inisialisasi pada class induk (Hero) tidak pernah dijalankan. Akibatnya, atribut dasar seperti nama, hp, dan attack_power tidak pernah dibuat/didaftarkan ke dalam objek Mage.  Peran super(): Menghubungkan class anak ke class induk untuk mengeksekusi constructor induk, sehingga data dasar diinisialisasi secara otomatis tanpa perlu ditulis ulang.  

Analisis 4
- Name Mangling (_Hero__hp): Memanggil hero1._Hero__hp tetap bisa menampilkan angka. Ini karena Python tidak mengunci variabel private secara ketat, melainkan hanya mengubah namanya secara otomatis (Name Mangling). Meskipun bisa diakses paksa, hal ini dilarang dalam standar coding yang baik karena merusak prinsip proteksi data.  

- Pentingnya Setter: Tanpa logika penyaring if/elif, perintah hero1.set_hp(-100) membuat HP menjadi bernilai negatif. Fungsi Setter sangat krusial sebagai pintu validasi agar data penting (seperti HP) selalu berada dalam rentang angka yang logis dan aman dari kecurangan/bug.  

```python
from abc import ABC, abstractmethod


# ==========================================
# LATIHAN 5: ABSTRACTION & INTERFACE (GameUnit)
# ==========================================
# Contract / Abstract Class Blueprint
class GameUnit(ABC):

    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass


# ==========================================
# LATIHAN 6: POLYMORPHISM (Hero & Turunannya)
# ==========================================
class Hero(GameUnit):

    def __init__(self, nama):
        self.nama = nama

    def serang(self, target=None):
        if target:
            print(f"Hero {self.nama} menebas {target}!")
        else:
            print("Hero menyerang dengan tangan kosong.")

    def info(self):
        print(f"Saya adalah Hero: {self.nama}")


class Mage(Hero):

    def serang(self, target=None):
        print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")


class Archer(Hero):

    def serang(self, target=None):
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")


class Fighter(Hero):

    def serang(self, target=None):
        print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")


# Tambahan class Healer untuk Uji Skalabilitas (Analisis 6.1)
class Healer(Hero):

    def serang(self, target=None):
        print(
            f"{self.nama} tidak menyerang, tapi menyembuhkan"
            " teman!"
        )


class Monster(GameUnit):

    def __init__(self, jenis):
        self.jenis = jenis

    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")

    def info(self):
        print(f"Saya adalah Monster: {self.jenis}")


# ==========================================
# MAIN PROGRAM & UJI COBA
# ==========================================
if __name__ == "__main__":
    print("--- Latihan 5: Abstraction & Interface ---")
    h = Hero("Alucard")
    m = Monster("Serigala")
    h.info()
    m.info()
    h.serang("Serigala")
    m.serang("Alucard")

    print("\n--- Latihan 6: Polymorphism dalam Loop ---")
    pasukan = [
        Mage("Eudora"),
        Archer("Miya"),
        Fighter("Zilong"),
        Mage("Gord"),
        Healer("Rafaela"),  # Class baru tanpa merusak struktur loop
    ]

    print("--- PERANG DIMULAI ---")
    for pahlawan in pasukan:
        pahlawan.serang()
```

Analisis 5
1. Melanggar Kontrak: Mengedit atau menghapus method serang dari Hero menyebabkan TypeError: Can't instantiate abstract class Hero with abstract method serang. Hal ini terjadi karena class turunan bertindak sebagai pelaksana "kontrak". Jika ada satu saja @abstractmethod yang lupa dibuat, class anak tersebut masih berstatus abstrak dan dilarang untuk dijadikan objek.  Mencetak Cetakan (GameUnit()):

2. Menjalankan unit = GameUnit() menghasilkan error karena class abstrak hanya berfungsi sebagai blueprint / kerangka dasar. Class ini ada untuk memaksakan standardisasi atribut/method pada seluruh class turunan, bukan untuk diinstansiasi menjadi objek nyata.

Analisis 6

1. Uji Skalabilitas: Saat class Healer ditambahkan, program berjalan lancar tanpa perlu mengubah kode loop for pahlawan in pasukan: sedikit pun. Keuntungan Polimorfisme adalah memberikan fleksibilitas tinggi bagi programmer untuk menambah jenis karakter atau fitur baru kapan saja tanpa mengganggu kode utama yang sudah berjalan.  

2. Konsistensi Penamaan: Mengubah method serang milik Archer menjadi tembak_panah menyebabkan AttributeError saat dilakukan perulangan. Polimorfisme sangat bergantung pada nama method yang seragam antar parent dan child class agar satu jenis perintah perulangan dapat dieksekusi secara universal.  

```python
from abc import ABC, abstractmethod


# ==========================================
# 1. ABSTRACTION (Abstract Class Parent)
# ==========================================
class KamarHotel(ABC):

    def __init__(self, nama_kamar, stok, harga_dasar):
        self.nama_kamar = nama_kamar
        # Encapsulation: Attribute dibuat Private
        self.__stok = 0
        self.__harga_dasar = 0

        # Validasi awal menggunakan Setter
        self.set_stok(stok)
        self.set_harga_dasar(harga_dasar)

    # Getter & Setter Stok
    def get_stok(self):
        return self.__stok

    def set_stok(self, jumlah):
        if jumlah < 0:
            print(
                f"Gagal update stok {self.nama_kamar}! Stok tidak boleh"
                f" negatif ({jumlah})."
            )
        else:
            self.__stok = jumlah
            print(
                f"Berhasil menambahkan stok {self.nama_kamar}:"
                f" {self.__stok} unit."
            )

    # Getter & Setter Harga Dasar
    def get_harga_dasar(self):
        return self.__harga_dasar

    def set_harga_dasar(self, harga):
        if harga >= 0:
            self.__harga_dasar = harga

    # Abstract Methods (Kontrak yang wajib di-override oleh child class)
    @abstractmethod
    def tampilkan_detail(self):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass


# ==========================================
# 2. INHERITANCE & POLYMORPHISM: Kamar Deluxe
# ==========================================
class KamarDeluxe(KamarHotel):

    def __init__(self, nama_kamar, stok, harga_dasar, fasilitas):
        super().__init__(nama_kamar, stok, harga_dasar)
        self.fasilitas = fasilitas

    # Pajak khusus Deluxe = 10%
    def hitung_pajak(self):
        return self.get_harga_dasar() * 0.10

    def hitung_harga_total(self, jumlah_malam):
        harga_per_malam = self.get_harga_dasar() + self.hitung_pajak()
        return harga_per_malam * jumlah_malam

    def tampilkan_detail(self):
        pajak = self.hitung_pajak()
        print(
            f"[DELUXE] {self.nama_kamar} | Fasilitas: {self.fasilitas}"
        )
        print(
            f"Harga Dasar/Malam: Rp {self.get_harga_dasar():,.0f} | Pajak"
            f" (10%): Rp {pajak:,.0f}".replace(",", ".")
        )


# ==========================================
# 3. INHERITANCE & POLYMORPHISM: Kamar Standard
# ==========================================
class KamarStandard(KamarHotel):

    def __init__(self, nama_kamar, stok, harga_dasar, kapasitas):
        super().__init__(nama_kamar, stok, harga_dasar)
        self.kapasitas = kapasitas

    # Pajak khusus Standard = 5%
    def hitung_pajak(self):
        return self.get_harga_dasar() * 0.05

    def hitung_harga_total(self, jumlah_malam):
        harga_per_malam = self.get_harga_dasar() + self.hitung_pajak()
        return harga_per_malam * jumlah_malam

    def tampilkan_detail(self):
        pajak = self.hitung_pajak()
        print(
            f"[STANDARD] {self.nama_kamar} | Kapasitas: {self.kapasitas}"
        )
        print(
            f"Harga Dasar/Malam: Rp {self.get_harga_dasar():,.0f} | Pajak"
            f" (5%): Rp {pajak:,.0f}".replace(",", ".")
        )


# ==========================================
# 4. POLYMORPHISM (Fungsi Transaksi)
# ==========================================
def proses_transaksi(daftar_pesanan):
    print("\nSTRUK PEMESANAN")
    total_tagihan = 0

    for idx, (kamar, malam) in enumerate(daftar_pesanan, start=1):
        print(f"{idx}. ", end="")
        kamar.tampilkan_detail()
        subtotal = kamar.hitung_harga_total(malam)
        total_tagihan += subtotal
        print(
            f"Menginap: {malam} malam | Subtotal: Rp"
            f" {subtotal:,.0f}\n".replace(",", ".")
        )

    print(
        f"TOTAL TAGIHAN: Rp {total_tagihan:,.0f}".replace(",", ".")
    )


# ==========================================
# MAIN PROGRAM (Sesuai User Story)
# ==========================================
if __name__ == "__main__":
    print("SETUP DATA KAMAR")

    # Admin membuat data kamar Deluxe
    deluxe1 = KamarDeluxe(
        "Kamar Deluxe Sea View", 10, 1500000, "Private Pool"
    )

    # Admin mencoba memasukkan stok negatif pada Kamar Standard (Akan Ditolak)
    standard1 = KamarStandard("Kamar Standard Superior", -5, 500000, "2 Orang")

    # Admin mengupdate stok Kamar Standard dengan angka positif
    standard1.set_stok(20)

    # Tamu memesan 2 malam Deluxe & 1 malam Standard
    pesanan = [(deluxe1, 2), (standard1, 1)]

    # Memproses pesanan dan mencetak struk
    proses_transaksi(pesanan)
```

Abstraction: Class KamarHotel mewajibkan method tampilkan_detail() dan hitung_harga_total() untuk diisi oleh class turunan. Class KamarHotel sendiri tidak bisa diinstansiasi langsung.

Encapsulation: Atribut __stok dan __harga_dasar bersifat private. Perubahan stok wajib melewati method set_stok() agar nilai negatif tidak bisa masuk.  

Inheritance: Class KamarDeluxe dan KamarStandard mewarisi seluruh atribut dasar dan setter/getter dari KamarHotel menggunakan fungsi super().__init__().

Polymorphism: Fungsi proses_transaksi() memproses list gabungan dari berbagai tipe kamar secara fleksibel. Setiap tipe kamar menghitung pajak dan menampilkan detail yang berbeda walaupun dipanggil menggunakan nama method yang sama.
